```python
from pathlib import Path
import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report,
    roc_auc_score, average_precision_score
)
```

```python
path = kagglehub.dataset_download("janiobachmann/bank-marketing-dataset")
filepath = Path(path) / Path("bank.csv")
data = pd.read_csv(filepath)
data
```

```python
data.info()
```

```python
sns.countplot(data=data, x="deposit")
plt.title("Distribution of target variable 'deposit'")
plt.show()
```

```python
data.describe().T
```

```python
numerical_features = ["age", "balance", "duration", "campaign", "pdays", "previous"]
data[numerical_features].hist(bins=15, figsize=(15, 10))
plt.suptitle("Distribution of Numerical Features")
plt.show()
```

```python
categorical_features = ["job", "marital", "education", "default", "housing", "loan", "contact", "day", "month", "poutcome"]
fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(15, 20))
for ax, feature in zip(axes.ravel(), categorical_features):
    sns.countplot(data=data, x=feature, hue="deposit", ax=ax)
    ax.set_title(f"{feature} vs deposit")
    ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

```python
missing_contact = (data["contact"] == "unknown").mean() * 100
missing_pdays = (data["pdays"] == -1).mean() * 100
missing_poutcome = (data["poutcome"] == "unknown").mean() * 100
missing_job = (data["job"] == "unknown").mean() * 100
missing_education = (data["education"] == "unknown").mean() * 100

pd.DataFrame(
    {
        "feature": ["contact", "pdays", "poutcome", "job", "education"],
        "missing_%": [missing_contact, missing_pdays, missing_poutcome, missing_job, missing_education],
    }
)
```

```python
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(15, 15))
for ax, feature in zip(axes.ravel(), numerical_features):
    sns.boxplot(data=data, x="deposit", y=feature, ax=ax)
    ax.set_title(f"{feature} vs deposit")
plt.tight_layout()
plt.show()
```

```python
x = data.drop(columns=["deposit", "duration", "pdays", "poutcome"])
y = data["deposit"]
y = y.map({"yes": 1, "no": 0})

x["month"] = x["month"].map({
    "jan": 1, "feb": 2, "mar": 3, "apr": 4, "may": 5, "jun": 6,
    "jul": 7, "aug": 8, "sep": 9, "oct": 10, "nov": 11, "dec": 12
})

binary_columns = ["default", "housing", "loan"]
for col in binary_columns:
    x[col] = x[col].replace({"yes": 1, "no": 0, "unknown": np.nan})
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
preprocessing = make_pipeline(
    ColumnTransformer(
        transformers=[
            ("num", make_pipeline(SimpleImputer(strategy="median"), PowerTransformer()),
             ["age", "balance", "campaign", "previous"]),
            ("cat", make_pipeline(SimpleImputer(strategy="most_frequent"), OneHotEncoder(handle_unknown="ignore")),
             ["marital", "contact", "job"])
        ],
        remainder="drop"
    )
)
```

```python
X_train_t = preprocessing.fit_transform(X_train)
feature_names = preprocessing.named_steps["columntransformer"].get_feature_names_out()
X_train_t = pd.DataFrame(X_train_t, columns=feature_names)
X_train_t.head()
```

```python
X_test_t = preprocessing.transform(X_test)
X_test_t = pd.DataFrame(X_test_t, columns=feature_names)
X_test_t.head()
```

```python
POS_LABEL = 1

log_reg = LogisticRegression(max_iter=2000)
param_grid_lr = {
    "C": [0.01, 0.1, 1, 10],
    "penalty": ["l2"],
    "solver": ["lbfgs"]
}
grid_lr = GridSearchCV(log_reg, param_grid_lr, scoring="f1", cv=5, n_jobs=-1)
grid_lr.fit(X_train_t, y_train)
best_lr = grid_lr.best_estimator_
best_lr, grid_lr.best_params_, grid_lr.best_score_
```

```python
rf = RandomForestClassifier(random_state=42)
param_grid_rf = {
    "n_estimators": [200, 400],
    "max_depth": [None, 8, 16],
    "min_samples_split": [2, 10],
    "min_samples_leaf": [1, 5]
}
grid_rf = GridSearchCV(rf, param_grid_rf, scoring="f1", cv=5, n_jobs=-1)
grid_rf.fit(X_train_t, y_train)
best_rf = grid_rf.best_estimator_
best_rf, grid_rf.best_params_, grid_rf.best_score_
```

```python
svc = SVC()
param_grid_svc = {
    "C": [0.1, 1, 10],
    "kernel": ["rbf", "linear"],
    "gamma": ["scale", "auto"]
}
grid_svc = GridSearchCV(svc, param_grid_svc, scoring="f1", cv=5, n_jobs=-1)
grid_svc.fit(X_train_t, y_train)
best_svc = grid_svc.best_estimator_
best_svc, grid_svc.best_params_, grid_svc.best_score_
```

```python
def evaluate_model(name, model, X_te, y_te, pos_label=POS_LABEL):
    y_pred = model.predict(X_te)

    proba = None
    score = None
    if hasattr(model, "predict_proba"):
        proba = model.predict_proba(X_te)[:, 1]
    elif hasattr(model, "decision_function"):
        score = model.decision_function(X_te)

    out = {
        "model": name,
        "test_accuracy": accuracy_score(y_te, y_pred),
        "test_precision": precision_score(y_te, y_pred, pos_label=pos_label, zero_division=0),
        "test_recall": recall_score(y_te, y_pred, pos_label=pos_label, zero_division=0),
        "test_f1": f1_score(y_te, y_pred, pos_label=pos_label, zero_division=0),
    }

    if proba is not None:
        out["test_roc_auc"] = roc_auc_score(y_te, proba)
        out["test_pr_auc"] = average_precision_score(y_te, proba)
    elif score is not None:
        out["test_roc_auc"] = roc_auc_score(y_te, score)
        out["test_pr_auc"] = average_precision_score(y_te, score)
    else:
        out["test_roc_auc"] = np.nan
        out["test_pr_auc"] = np.nan

    cm = confusion_matrix(y_te, y_pred)
    ConfusionMatrixDisplay(cm).plot()
    plt.title(f"Confusion Matrix - {name}")
    plt.show()

    print(classification_report(y_te, y_pred, zero_division=0))
    return out
```

```python
results = []
results.append({**evaluate_model("Logistic Regression", best_lr, X_test_t, y_test), "cv_best_f1": grid_lr.best_score_})
results.append({**evaluate_model("Random Forest", best_rf, X_test_t, y_test), "cv_best_f1": grid_rf.best_score_})
results.append({**evaluate_model("SVC", best_svc, X_test_t, y_test), "cv_best_f1": grid_svc.best_score_})

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="test_f1", ascending=False).reset_index(drop=True)
results_df["f1_gap_cv_minus_test"] = results_df["cv_best_f1"] - results_df["test_f1"]
results_df
```

```python
best_name = results_df.iloc[0]["model"]
best_model = {
    "Logistic Regression": best_lr,
    "Random Forest": best_rf,
    "SVC": best_svc
}[best_name]
print("Selected best model:", best_name)
```

```python
best_row = results_df.iloc[0].to_dict()
print("=== Summary ===")
print(f"Best model: {best_row['model']}")
print(f"Test F1: {best_row['test_f1']:.4f} | Precision: {best_row['test_precision']:.4f} | Recall: {best_row['test_recall']:.4f} | Accuracy: {best_row['test_accuracy']:.4f}")
if not pd.isna(best_row.get("test_pr_auc", np.nan)):
    print(f"PR-AUC: {best_row['test_pr_auc']:.4f} | ROC-AUC: {best_row['test_roc_auc']:.4f}")
print(f"CV best F1: {best_row['cv_best_f1']:.4f} | CV-Test F1 gap: {best_row['f1_gap_cv_minus_test']:.4f}")
```
